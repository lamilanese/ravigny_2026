```python
from PIL import Image

def png_to_ico(png_path, ico_path, sizes=None):
    """
    Convert a PNG image to an ICO file.

    :param png_path: Path to source PNG
    :param ico_path: Path to output ICO
    :param sizes: List of (width, height) tuples for icon sizes
    """
    if sizes is None:
        sizes = [(16, 16), (32, 32), (48, 48), (64, 64), (128, 128), (256, 256)]

    img = Image.open(png_path)
    img.save(ico_path, format="ICO", sizes=sizes)

if __name__ == "__main__":
    png_to_ico("icon.png", "icon.ico")
```
